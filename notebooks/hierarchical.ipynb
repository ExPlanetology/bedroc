```python
import logging

from bedroc import debug_logger
from bedroc.hierarchical import Analyzer, SyntheticDataGenerator, hierarchical_difference_model

logger = debug_logger()
logger.setLevel(logging.INFO)
```

# Hierarchical Bayesian analysis for multi-feature data

Set a random seed for reproducibility

```python
RANDOM_SEED: int = 123
```

Generate synthetic data

```python
data_generator: SyntheticDataGenerator = SyntheticDataGenerator(random_seed=RANDOM_SEED)
data_generator.generate()
```

Plot the synthetic data

```python
data_generator.plot()
```

Run the hierarchical difference model with the synthetic data

```python
model, idata = hierarchical_difference_model(data_generator.X_A, data_generator.X_B)
```

Create a data analyzer

```python
analyzer = Analyzer(idata)
```

Generate plots

```python
_ = analyzer.plot_posterior_differences()
```

```python
_ = analyzer.plot_posterior_effect_size()
```
